## Objective
Collect publicly available SpaceX launch data using the SpaceX REST API and convert it into a pandas DataFrame for further cleaning, exploratory analysis, and machine learning.

## 1. Import Libraries
The requests library is used to access the API, while pandas is used to organize the returned data.

```python
import requests
import pandas as pd
import json
from datetime import datetime
```

## 2. Define the API Endpoint
The SpaceX API provides historical launch records in JSON format.

```python
api_url = 'https://api.spacexdata.com/v4/launches'
response = requests.get(api_url, timeout=30)
response.raise_for_status()
launch_data = response.json()

print(f'Total launch records collected: {len(launch_data)}')
```

## 3. Inspect the Returned Data
Display one record to understand the structure of the API response.

```python
print(json.dumps(launch_data[0], indent=2)[:5000])
```

## 4. Convert the JSON Data into a DataFrame
The list of launch records is converted into a tabular pandas DataFrame.

```python
launch_df = pd.json_normalize(launch_data)
print('DataFrame shape:', launch_df.shape)
launch_df.head()
```

## 5. Inspect Columns and Missing Values

```python
print('Column names:')
for column in launch_df.columns:
    print(column)

missing_values = launch_df.isnull().sum().sort_values(ascending=False)
missing_values.head(20)
```

## 6. Select Initial Fields for the Project
The following fields are useful for studying launch and landing patterns. Some fields may not exist in every API version, so the code checks availability before selecting them.

```python
candidate_columns = [
    'id', 'name', 'flight_number', 'date_utc', 'success',
    'details', 'rocket', 'launchpad', 'payloads', 'cores',
    'orbit', 'payload_mass_kg', 'payload_mass_lbs'
]

available_columns = [c for c in candidate_columns if c in launch_df.columns]
selected_launch_df = launch_df[available_columns].copy()

print('Selected columns:')
print(available_columns)
selected_launch_df.head()
```

## 7. Save the Raw API Data
The collected data is saved locally so that it can be reused during the data-wrangling stage.

```python
launch_df.to_csv('spacex_launches_raw.csv', index=False)
selected_launch_df.to_csv('spacex_launches_selected.csv', index=False)
print('Raw and selected datasets saved successfully.')
```

## Conclusion
The SpaceX launch records were collected from the public API and converted into a pandas DataFrame. The data is now ready for cleaning, feature engineering, exploratory data analysis, visualization, SQL analysis, and predictive modelling.
